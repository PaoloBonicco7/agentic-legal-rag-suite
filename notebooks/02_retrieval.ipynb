# Retrieval diagnostics
Compare dense, sparse, hybrid and bounded graph retrieval without answer generation or judging. Article success/MRR are incomplete-reference proxies, not legal correctness. Core behavior: specs 04, 05 and 08.


```python
from pathlib import Path

PROJECT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
CORPUS = PROJECT / "data/corpora/validated_20261011"
INDEX = PROJECT / "data/indexes/validated-bge_20261011"
BENCHMARK = PROJECT / "data/evaluation"
DIAGNOSTICS = PROJECT / "data/diagnostics/validated_20261011/retrieval"
NEW_DIAGNOSTICS = PROJECT / "data/diagnostics/notebook-retrieval"
RUN_RETRIEVAL = False
```

## 1. Optional local diagnostic run
Enable only after indexing completes and no other process holds the local index. NEW_DIAGNOSTICS is a new output directory; the delivered diagnostics remain unchanged. This loads cached model weights and runs query embeddings. Remote embedding indexes may call their provider; use a local index for this demonstration. Gold references are evaluated after retrieval.


```python
if RUN_RETRIEVAL:
    from legal_rag.cli import main

    exit_code = main(["diagnose", "--index", str(INDEX), "--corpus", str(CORPUS),
        "--benchmark", str(BENCHMARK), "--output", str(NEW_DIAGNOSTICS), "--local-files-only"])
    print({"exit_code": exit_code})
    if exit_code == 0:
        DIAGNOSTICS = NEW_DIAGNOSTICS
```

## 2. Inspect saved results
Inspection below opens no index and loads no model. Keep original candidate pool, selected candidates and packed context distinct.


```python
from legal_rag.artifacts import read_json
from legal_rag.evaluation import load_diagnostics

if (DIAGNOSTICS / "manifest.json").is_file():
    print(read_json(DIAGNOSTICS / "metrics.json"))
    records = load_diagnostics(DIAGNOSTICS)
    print({"qid": records[0].qid, "profile": records[0].answer.profile,
           "retrieval_proxies": records[0].retrieval_proxies})
else:
    print("No standalone diagnostics at this path; inspect saved evaluation runs in notebook 03.")
```
